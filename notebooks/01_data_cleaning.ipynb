```python
#Filter for entry-level friendly roles
#EN = Entry, MI = Mid-level
entry_level_mask = df['experience_level'].isin(['EN', 'MI'])
df_entry = df[entry_level_mask].copy()

print(f"Original: {len(df)} rows")
print(f"Entry/Mid friendly: {len(df_entry)} rows")
```

```python
df_clean = df_entry[df_entry['salary_in_usd'].notna()].copy()

```

```python
output_path = "E:\\f\\Portfolio Projects\\1\\Data-Job-Market-Intelligence\\data\\salaries_cleaned.csv"
df_clean.to_csv(output_path, index=False)
print(f"Cleaned data saved to {output_path}")
print(f"Shape: {df_clean.shape}")
```
